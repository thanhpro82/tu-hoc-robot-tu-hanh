## 4. Thử thách

Chỉ giữ lại một loại sai số mỗi lần, xem loại nào làm robot lệch nhiều nhất sau 3 vòng.

```python
import loi_giai
kh = loi_giai.ke_hoach_vuong(1.0, 0.5, 1.0)
for ten in ["v_ty_le", "w_ty_le", "w_lech", "nhieu"]:
    chi_mot = {"v_ty_le": 1.0, "w_ty_le": 1.0, "w_lech": 0.0, "nhieu": 0.0}
    chi_mot[ten] = goc.ROBOT_THAT[ten]
    _, lech = goc.chay(kh, so_vong=3, sai_so=chi_mot, seed=3)
    print(f"chỉ có {ten:8s} = {goc.ROBOT_THAT[ten]:+.2f}: lệch sau 3 vòng {lech[-1] * 100:5.1f} cm")
```
