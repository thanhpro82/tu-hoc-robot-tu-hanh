## 4. Thử thách

Đếm tham số từng lớp của SimpleCNN, so với mạng nối đầy đủ 1024 → 256 → 4.

```python
import loi_giai
m = loi_giai.SimpleCNN()
for ten, lop in m.named_children():
    n = sum(p.numel() for p in lop.parameters())
    print(f"{ten:6s}: {n:6d} tham số")
print("tổng SimpleCNN:", sum(p.numel() for p in m.parameters()))
print("mạng nối đầy đủ 1024→256→4:", (1024 + 1) * 256 + (256 + 1) * 4)
```
