| **2** | `van_toc_ngang(path)` | vận tốc ngang lớn nhất trên cả đường đi |

## 4. Thử thách

Đổi góc lái tối đa rồi xem bán kính quay và quãng lùi thay đổi thế nào.

```python
phi_max_do = 45      # thử 20, 30, 45
L = 0.30
R = L / math.tan(math.radians(phi_max_do))
alpha = math.acos(1 - 0.6 / (2 * R))
print(f"φ_max = {phi_max_do}°: R = {R:.2f} m, mỗi cung {math.degrees(alpha):.0f}°, xe phải lùi {2 * R * math.sin(alpha):.2f} m theo phương dọc")
```
