## 4. Thử thách

Đổi độ mịn σ của lọc Gauss: số cạnh giả và độ chính xác của vạch thay đổi thế nào?

```python
import loi_giai
from kiem_tra import canh_voi
for sigma in [0.5, 1.0, 2.0, 4.0]:
    goc.SIGMA = sigma
    e = canh_voi(loi_giai.do_doc, goc.ANH, True)
    n, g = goc.thong_ke(e)
    dinh = goc.tim_dinh(*goc.hough(e)) if n else []
    vach = f"vạch α = {np.mean([d[0] for d in dinh]):.1f}° (thật 23.4°)" if dinh else "không còn cạnh nào vượt ngưỡng"
    print(f"σ = {sigma}: {n:5d} pixel cạnh, {g / max(n, 1) * 100:3.0f}% giả, {vach}")
goc.SIGMA = 2.0
```
