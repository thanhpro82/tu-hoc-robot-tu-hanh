## 4. Thử thách

Đổi lệnh rồi chạy lại. Thử: `v` tăng dần theo thời gian (xoắn ốc), hoặc cộng 0,05 rad/s vào ω (bánh lệch).

```python
import matplotlib.pyplot as plt
v = 0.5
omega = lambda t: 1.0 + 0.05          # thử đổi: lambda t: 0.3 * t, ...
x, y, th, xs, ys = 0.0, 0.0, 0.0, [], []
for i in range(1500):
    x, y, th = goc.step(x, y, th, v, omega(i * goc.dt))
    xs.append(x); ys.append(y)
plt.plot(xs, ys); plt.axis("equal"); plt.grid(True); plt.show()
```
